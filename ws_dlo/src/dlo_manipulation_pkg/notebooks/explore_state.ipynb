```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D  # Needed for 3D projections
import pandas as pd
```

```python
data_path = '/home/jessica/shape_control_DLO_2/data/train_data/2D/state_0.npy'
state = np.load(data_path)
```

```python
# 1. Print overall shape and type
print('Shape:', state.shape)
print('Data type:', state.dtype)
```

```python
# 2. Show the first 5 rows
print('First five rows:')
print(state[:5])
```

```python
# Set which row to plot
row_num = 200  # Change this to plot a different row
```

```python
# Extract feature points (columns 1-30 → 10 points each with x/y/z)
fps_row = state[row_num, 1:31]  # shape (30,)
fps = fps_row.reshape(10, 3)    # shape (10, 3); columns: x, y, z

plt.figure(figsize=(7, 6))
plt.scatter(fps[:, 0], fps[:, 1], color='blue')  # plot x vs y
for i in range(10):
    plt.text(fps[i, 1], fps[i, 2], str(i + 1), fontsize=12,
             ha='right', va='bottom', color='darkred')
plt.plot(fps[:, 1], fps[:, 2], 'k--', alpha=0.5)
plt.xlabel('X')
plt.ylabel('Y')
plt.title(f'2D Feature Points in Y-Z plane (Row {row_num})')
plt.gca().set_aspect('equal')
plt.grid(True)
plt.show()
```

```python
def print_feature_points_table(state, row_num):
    # Extract feature points for the specified row
    fps_row = state[row_num, 1:31]  # columns 1 to 30 contain feature points (10 * 3)
    fps = fps_row.reshape(10, 3)
    
    # Create DataFrame for clean display
    df = pd.DataFrame(fps, columns=['X', 'Y', 'Z'])
    df.index = [f'Point {i+1}' for i in range(10)]
    print(f'Feature Point Coordinates for Row {row_num}:\n')
    print(df)
```

```python
print_feature_points_table(state, row_num)
```
